```python
target = "smoking"
drop_cols = ["ID", "oral"]
cat_cols = ["gender", "tartar", "dental caries", "hearing(left)", "hearing(right)"]
ord_cols = ["Urine protein"]
num_cols = [c for c in df.columns
            if c not in cat_cols + ord_cols + drop_cols + [target]]

print(len(num_cols), num_cols)   
df_num = df[num_cols]

for col in df_num.columns:
    df_num.boxplot(column=col)
    plt.show()
```

- provavéis de erros de medição: eyesigth (left and rigth), LDL e HDL 
- valores reais mas assimétricos: serum creatinine, triglyceride, fasting blood sugar, Gtp, ALT, AST  

- efetuar limpeza de outliers e consequentemente injetação de elementos

```python
df_num.hist(bins=50, figsize=(20,15));
plt.show()
```

```python
corr_cols = num_cols + [target]          # as 18 numéricas + smoking
corr_coef = df[corr_cols].corr()
# print(corr_coef)
corr_coef["smoking"].sort_values(ascending=False)
```

```python
plt.figure(figsize=(14, 11))
sns.heatmap(corr_coef, cmap="coolwarm", center=0, annot=True, fmt=".2f",
            annot_kws={"size": 7})
plt.show()
```
